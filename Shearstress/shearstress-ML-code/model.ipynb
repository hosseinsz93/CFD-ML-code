```python
import tensorflow as tf

tf.keras.backend.set_floatx('float32')


class ConvNet(tf.keras.Model):
    def __init__(self):
        super().__init__()
        self.layer1 = tf.keras.Sequential([
            tf.keras.layers.ZeroPadding2D(padding=((3, 3), (1, 1))),
            tf.keras.layers.Conv2D(8, (7, 3), strides=(1, 1), padding='valid', use_bias=False),
            tf.keras.layers.LeakyReLU(0.1),
        ])

        self.layer2 = tf.keras.Sequential([
            tf.keras.layers.ZeroPadding2D(padding=((3, 3), (1, 1))),
            tf.keras.layers.Conv2D(16, (9, 3), strides=(5, 2), padding='valid', use_bias=False),
            tf.keras.layers.LeakyReLU(0.1),
        ])

        self.layer3 = tf.keras.Sequential([
            tf.keras.layers.ZeroPadding2D(padding=((3, 3), (1, 1))),
            tf.keras.layers.Conv2D(32, (7, 3), strides=(4, 2), padding='valid', use_bias=False),
            tf.keras.layers.LeakyReLU(0.1),
        ])

        self.layer4 = tf.keras.Sequential([
            tf.keras.layers.ZeroPadding2D(padding=((3, 3), (1, 1))),
            tf.keras.layers.Conv2D(64, (7, 3), strides=(4, 2), padding='valid', use_bias=False),
            tf.keras.layers.LeakyReLU(0.1),
        ])

        self.layer5 = tf.keras.Sequential([
            tf.keras.layers.Conv2DTranspose(32, (6, 4), strides=(4, 2), padding='valid', use_bias=False),
            tf.keras.layers.Cropping2D(cropping=((1, 1), (1, 1))),
            tf.keras.layers.PReLU(shared_axes=[1, 2]),
        ])

        self.layer6 = tf.keras.Sequential([
            tf.keras.layers.Conv2DTranspose(16, (6, 4), strides=(4, 2), padding='valid', use_bias=False),
            tf.keras.layers.Cropping2D(cropping=((1, 1), (1, 1))),
            tf.keras.layers.PReLU(shared_axes=[1, 2]),
        ])

        self.layer7 = tf.keras.Sequential([
            tf.keras.layers.Conv2DTranspose(8, (9, 4), strides=(5, 2), padding='valid', use_bias=False),
            tf.keras.layers.Cropping2D(cropping=((2, 2), (1, 1))),
            tf.keras.layers.PReLU(shared_axes=[1, 2]),
        ])

        self.output_layer = tf.keras.Sequential([
            tf.keras.layers.ZeroPadding2D(padding=((3, 3), (1, 1))),
            tf.keras.layers.Conv2D(1, (7, 3), strides=(1, 1), padding='valid'),
            tf.keras.layers.PReLU(shared_axes=[1, 2]),
        ])

    def call(self, inputs, training=False):
        x = self.layer1(inputs)
        x = self.layer2(x)
        x = self.layer3(x)
        x = self.layer4(x)
        x = self.layer5(x)
        x = self.layer6(x)
        x = self.layer7(x)
        return self.output_layer(x)
```
